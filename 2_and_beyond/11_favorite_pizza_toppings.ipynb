```python
# Exercise 11: Favorite Pizza Toppings

import matplotlib.pyplot as plt
import numpy as np

toppings = np.array(['Pepperoni', 'Cheese', 'Mushroom', 'Pineapple', 'Prosciutto', 'Artichoke'])
orders = np.array([40, 25, 20, 15, 4, 5])

plt.pie(orders, labels=toppings, autopct='%1.1f%%', explode=[0.1, 0, 0, 0, 0, 0])
plt.title('Favourite Pizza Toppings')
plt.show()
```
