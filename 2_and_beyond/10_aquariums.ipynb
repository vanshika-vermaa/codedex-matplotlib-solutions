```python
# Exercise 10: Aquariums

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

df = pd.DataFrame({
  'aquarium': ['Monterrey Bay', 'Georgia Aquarium', 'Shedd Aquarium', 'National Aquarium'],
  'shark': [5, 4, 4, 2],
  'dolphin': [8, 12, 10, 5],
  'octopus': [7, 20, 8, 4],
  'sea turtle': [12, 5, 6, 3],
  'jellyfish': [30, 25, 20, 15]
})

plt.figure()

# Amount of sharks at each acquarium
plt.bar(df['aquarium'], df['shark'])
plt.title('Number of Sharks at each Acquarium')
plt.xlabel('Acquarium')
plt.ylabel('Number of Sharks')
# plt.tight_layout()     # can use this to make things appear neatly
plt.show()
```
