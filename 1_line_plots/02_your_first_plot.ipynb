```python
# Exercise 02: Your First Plot

import matplotlib.pyplot as plt
import numpy as np

# Array for days
days = np.array(['Sunday', 'Monday', 'Tuesday', 'Wednesday', 'Thursday','Friday', 'Saturday'])

# Screen time in minutes for each day
screen_time = np.array([260, 483, 390, 375, 489, 370, 349])

# Create the plot: days vs screen_time
plt.plot(days, screen_time)

# Display the plot
plt.show()
```
