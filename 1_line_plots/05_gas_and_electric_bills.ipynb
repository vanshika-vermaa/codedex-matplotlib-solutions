```python
# Exercise 05: Gas & Electric Bills

import matplotlib.pyplot as plt
import numpy as np

months = np.array(['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'])
gas = np.array([1200, 1150, 1300, 1250, 1400, 1350, 1500, 1450, 1380, 1320, 1280, 1220])
electric = np.array([900, 950, 1000, 1100, 1200, 1250, 1300, 1280, 1150, 1050, 980, 940])

plt.figure()
plt.plot(months, gas, label='Gas Bills')
plt.plot(months, electric, label='Electric Bills')

plt.title('Gas & Electric Bills')
plt.xlabel('Months')
plt.ylabel('Cost')

plt.legend()
plt.show()
```
