```python
# Exercise 06: Sports Rivalries

import matplotlib.pyplot as plt
import numpy as np

games = np.array([1, 2, 3, 4, 5])

# Points for the IPL teams
rcb_points = np.array([200, 187, 175, 190, 205])
srh_points = np.array([170, 180, 195, 160, 175])
rr_points = np.array([165, 190, 185, 200, 178])
dc_points = np.array([178, 201, 165, 189, 210])
csk_points = np.array([155, 172, 198, 160, 185])

plt.figure()
plt.plot(games, rcb_points, color='red', label='RCB')
plt.plot(games, srh_points, color='orange', label='SRH')
plt.plot(games, rr_points, color='pink', label='RR')
plt.plot(games, dc_points, color='blue', label='DC')
plt.plot(games, csk_points, color='yellow', label='CSK')

plt.title('Sports Rivalries')
plt.xlabel('Number of Games')
plt.ylabel('Runs Scored')

plt.legend()
plt.show()
```
