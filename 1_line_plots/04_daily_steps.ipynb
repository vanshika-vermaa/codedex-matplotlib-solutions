```python
# Exercise 04: Daily Steps

import matplotlib.pyplot as plt
import numpy as np

# List of days
days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

# Array of steps per day
steps = np.array([400, 553, 387, 620, 229, 345, 460])

# Create the plot
plt.figure()
plt.plot(days, steps)

# Give title to the plot
plt.title('Steps taken each day')

# Label the x and y axes
plt.xlabel('Days of the week')
plt.ylabel('Number of steps')

# Display the plot
plt.show()
```
